# Explore Rodent Reference Clips

This notebook explores the structure of the rodent reference clips h5 file.

```python
import h5py
import yaml
import numpy as np
```

```python
# Path to the reference clips file
h5_path = "../data/rodent/rodent_reference_clips.h5"
```

```python
def print_h5_structure(h5_file, indent=0):
    """Recursively print the structure of an h5 file."""
    for key in h5_file.keys():
        item = h5_file[key]
        prefix = "  " * indent
        if isinstance(item, h5py.Group):
            print(f"{prefix}{key}/ (Group)")
            print_h5_structure(item, indent + 1)
        elif isinstance(item, h5py.Dataset):
            print(f"{prefix}{key}: shape={item.shape}, dtype={item.dtype}")
        else:
            print(f"{prefix}{key}: {type(item)}")
```

```python
# Open and explore the h5 file
with h5py.File(h5_path, "r") as fid:
    print("=" * 60)
    print("H5 FILE STRUCTURE")
    print("=" * 60)
    print_h5_structure(fid)
    print()
```

```python
# Load and parse the config
with h5py.File(h5_path, "r") as fid:
    config = yaml.safe_load(fid["config"][()])

print("=" * 60)
print("CONFIG")
print("=" * 60)
for key, value in config.items():
    if isinstance(value, dict):
        print(f"{key}:")
        for k, v in value.items():
            if isinstance(v, list) and len(v) > 5:
                print(f"  {k}: [{len(v)} items]")
            else:
                print(f"  {k}: {v}")
    else:
        print(f"{key}: {value}")
```

```python
# Calculate number of clips
with h5py.File(h5_path, "r") as fid:
    qpos_shape = fid["qpos"].shape
    qvel_shape = fid["qvel"].shape
    xpos_shape = fid["xpos"].shape
    xquat_shape = fid["xquat"].shape

    # Get clip length from config
    config = yaml.safe_load(fid["config"][()])

    # Names
    names_qpos = fid["names_qpos"][()].astype(str)
    names_xpos = fid["names_xpos"][()].astype(str)

print("\n" + "=" * 60)
print("DATA ARRAYS")
print("=" * 60)
print(f"qpos shape: {qpos_shape}")
print(f"qvel shape: {qvel_shape}")
print(f"xpos shape: {xpos_shape}")
print(f"xquat shape: {xquat_shape}")

print(f"\nnames_qpos ({len(names_qpos)} entries): {list(names_qpos)}")
print(f"\nnames_xpos ({len(names_xpos)} entries): {list(names_xpos)}")
```

```python
# Calculate number of clips based on standard clip length
# The default clip_length in the imitation config is 250 frames
CLIP_LENGTH = 250

total_frames = qpos_shape[0]
n_clips = total_frames // CLIP_LENGTH

print("\n" + "=" * 60)
print("CLIP INFORMATION")
print("=" * 60)
print(f"Total frames: {total_frames}")
print(f"Clip length: {CLIP_LENGTH} frames")
print(f"Number of clips: {n_clips}")

# Verify with config snips_order if available
if "model" in config and "snips_order" in config["model"]:
    snips_order = config["model"]["snips_order"]
    print(f"\nClips from config (snips_order): {len(snips_order)} clips")
    print(f"Clip names (first 10): {snips_order[:10]}")
    if len(snips_order) > 10:
        print(f"... and {len(snips_order) - 10} more")
```

```python
# Show behavior types if available
import re

if "model" in config and "snips_order" in config["model"]:
    pattern = r"([A-Za-z]+)_\d+(\.p)?"
    behavior_counts = {}
    for fn in config["model"]["snips_order"]:
        m = re.search(pattern, fn)
        if m:
            behavior = m.group(1)
            behavior_counts[behavior] = behavior_counts.get(behavior, 0) + 1

    print("\n" + "=" * 60)
    print("BEHAVIOR TYPES")
    print("=" * 60)
    for behavior, count in sorted(behavior_counts.items(), key=lambda x: -x[1]):
        print(f"  {behavior}: {count} clips")
```

```python
# Calculate percentage of clips with "walk" in the name (case-insensitive)
if "model" in config and "snips_order" in config["model"]:
    snips_order = config["model"]["snips_order"]
    walk_clips = [name for name in snips_order if "walk" in name.lower()]
    walk_percentage = len(walk_clips) / len(snips_order) * 100

    print("=" * 60)
    print("WALK CLIPS")
    print("=" * 60)
    print(f"Total clips: {len(snips_order)}")
    print(f"Clips with 'walk' in name: {len(walk_clips)}")
    print(f"Percentage: {walk_percentage:.2f}%")
```

```python
# Print qpos of clip index 0 for the first 5 time steps
clip_index = 0
n_timesteps = 5

with h5py.File(h5_path, "r") as fid:
    qpos = fid["qpos"][:]

# Calculate start and end indices for clip 0
start_idx = clip_index * CLIP_LENGTH
end_idx = start_idx + n_timesteps

clip_qpos = qpos[start_idx:end_idx]

print("=" * 60)
print(f"QPOS FOR CLIP {clip_index} (first {n_timesteps} timesteps)")
print("=" * 60)
print(f"Shape: {clip_qpos.shape}")
print(f"\nClip name: {config['model']['snips_order'][clip_index]}")
print(f"\nqpos values:")
print(clip_qpos)
```

```python
# Find qpos indices that are 0 for all 5 time steps
zero_indices = np.where(np.all(clip_qpos == 0, axis=0))[0]

print("=" * 60)
print("QPOS INDICES THAT ARE ZERO FOR ALL 5 TIMESTEPS")
print("=" * 60)
print(f"Number of zero indices: {len(zero_indices)}")
print(f"Indices: {zero_indices.tolist()}")
print(f"\nCorresponding joint names:")
for idx in zero_indices:
    print(f"  {idx}: {names_qpos[idx]}")
```
