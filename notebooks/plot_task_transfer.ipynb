```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the data
df = pd.read_csv('../results/maintain_velocity_return.csv')

# Normalize returns by dividing by 2000
df['Prior-Decoder'] = df['Prior-Decoder'] / 2000
df['Decoder-Only'] = df['Decoder-Only'] / 2000

# Plot
fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(df['Speed'], df['Prior-Decoder'], marker='o', label='Prior-Decoder', color='blue')
ax.plot(df['Speed'], df['Decoder-Only'], marker='s', label='Decoder-Only', color='red')

ax.set_xlabel('Target Speed', fontsize=20)
ax.set_ylabel('Mean Normalized Return', fontsize=20)
ax.legend(fontsize=14)
ax.grid(True, alpha=0.3)
ax.tick_params(labelsize=16)
ax.set_title('Mean Normalized Return vs Speed', fontsize=20)
fig.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the bowl escape data
df = pd.read_csv('../results/bowl_escape.csv')

# Clean column names
df.columns = df.columns.str.strip()

# Plot
fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(df['num_steps'], df.iloc[:, 1], marker='o', label='Prior-Decoder', color='blue')
ax.plot(df['num_steps'], df.iloc[:, 2], marker='s', label='Decoder-Only', color='red')

ax.set_xlabel('Simulation Steps', fontsize=20)
ax.set_ylabel('Return', fontsize=20)
ax.legend(fontsize=14)
ax.grid(True, alpha=0.3)
ax.tick_params(labelsize=16)
ax.set_title('Bowl Escape', fontsize=20)
fig.tight_layout()
plt.show()
```
